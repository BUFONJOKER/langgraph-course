```python
def run_workflow(prompt: str):
    workflow.invoke(input={'messages':[('user', prompt)], 'summary':''}, config=config)
```

```python
def get_state():
    state = workflow.get_state(config)
    msgs = state.values['messages']
    summary = state.values['summary']
    return msgs, summary
```
